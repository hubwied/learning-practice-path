```python
class Laptop:
    
    brand = 'Apple'
    _usdpln = 4.2
    __trade_margin = 0.2
    
    def __init__(self,net_price):
        self.net_price = net_price
        
    def convert_to_pln(price):
        return price * Laptop._usdpln + (1+ Laptop._Laptop__trade_margin)
    
    def display_price(self):
        print(Laptop.convert_to_pln(self.net_price))
```

```python
laptop = Laptop(1000)
laptop.display_price()
```
